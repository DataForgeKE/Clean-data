```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\1040 G3\Downloads\messy_sales - Sheet1.csv")
```

First ten rows

```python
df.head(10)
```

Clean columns

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(' ','')
)
```

Clean date column

```python
df['Date'] = pd.to_datetime(df['Date'],format='mixed')
```

Clean region, product, category

```python
for col in ['Region','Product', 'Category']:
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

Clean category

```python
df['Category'] = (
    df['Category']
    .str.replace('Electronicstronics','Electronics')
    .str.replace('Gadjetss','Gadgets')
    .str.replace('Gadjets','Gadgets')
)
```

Clean units sold


```python
df['UnitsSold'] = (
    df['UnitsSold']
    # .astype[str]
    .str.replace('thirty','30',regex=True)
    .str.replace('twenty','20', regex=True)
)

df['UnitsSold'] = pd.to_numeric(df['UnitsSold'],errors='coerce')
```

Clean unit price

```python
df['UnitPrice'] = (
    df['UnitPrice']
    .astype(str)
    .str.replace('USD','')
    .str.replace('$','')
    .astype(float)
)
```

Clean profit margin

```python
df['ProfitMargin(%)'] = (
    df['ProfitMargin(%)']
    .astype(str)
    .str.replace('%','')
    .str.replace('percent','')
    .str.strip()
    .astype(float)
)
df['ProfitMargin(%)'] = np.where(df['ProfitMargin(%)']<1, df['ProfitMargin(%)']*100,df['ProfitMargin(%)'])
```

Clean total sales

```python
df['TotalSales'] = np.multiply(df['UnitPrice'],df['UnitsSold'])
```
